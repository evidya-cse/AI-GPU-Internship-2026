```python
import torch
import time
import psutil
from transformers import pipeline, AutoTokenizer, AutoModel

print("=" * 60)
print("HUGGING FACE AI WORKSTATION")
print("=" * 60)

device = "cuda" if torch.cuda.is_available() else "cpu"
print("\nDevice:", device)

text = "Artificial Intelligence is transforming education and cybersecurity."

print("\nTOKENIZATION")
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")
tokens = tokenizer.tokenize(text)
print("Tokens:", tokens)
print("Token Count:", len(tokens))

print("\nEMBEDDINGS")
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
model = AutoModel.from_pretrained("bert-base-uncased")
inputs = bert_tokenizer(text, return_tensors="pt")
with torch.no_grad():
    outputs = model(**inputs)
embedding = outputs.last_hidden_state
print("Embedding Shape:", embedding.shape)

print("\nTEXT GENERATION")
generator = pipeline("text-generation", model="distilgpt2")
result = generator("Machine Learning is", max_new_tokens=50)
print(result[0]["generated_text"])

print("\nSENTIMENT ANALYSIS")
sentiment = pipeline("sentiment-analysis")
reviews = ["This workshop is amazing.", "The session was boring."]
for review in reviews:
    print(sentiment(review))

print("\nSUMMARIZATION")
summary_prompt = """
Summarize this text in one short sentence:
Artificial Intelligence is a branch of computer science focused on creating systems capable of performing tasks that typically require human intelligence.
Summary:
"""
summary = generator(summary_prompt, max_new_tokens=40)
print(summary[0]["generated_text"])

print("\nTRANSLATION")
translation_prompt = """
Translate this sentence from English to French:
Artificial Intelligence is transforming education.
French:
"""
translated = generator(translation_prompt, max_new_tokens=40)
print(translated[0]["generated_text"])

print("\nBENCHMARK")
start = time.time()
output = generator("Explain Transformers in AI.", max_new_tokens=100)
end = time.time()
generated_text = output[0]["generated_text"]
tokens_generated = len(generated_text.split())
throughput = tokens_generated / (end - start)
print("Time:", round(end - start, 2), "sec")
print("Tokens/sec:", round(throughput, 2))

print("\nMEMORY")
process = psutil.Process()
memory = process.memory_info().rss / 1024**2
print("RAM Usage:", round(memory, 2), "MB")

if torch.cuda.is_available():
    gpu_memory = torch.cuda.memory_allocated() / 1024**3
    print("GPU Memory:", round(gpu_memory, 2), "GB")

print("\nExecution Complete")
```
